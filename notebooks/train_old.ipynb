# YOLOv8 Training Notebook

This notebook demonstrates how to train a YOLOv8 model using the Ultralytics API in a Jupyter environment.

## 1. Install Dependencies

```python
import os
import sys
from ultralytics import YOLO  # ultralytics>=8.0
import yaml
from pathlib import Path
from random import sample

# Add project root to system path (for relative imports to work)
project_root = os.path.abspath(os.path.join(os.getcwd(), ".."))
if project_root not in sys.path:
    sys.path.insert(0, project_root)

from src import config
```

## 2. Prepare Data Configuration (`data.yaml`)
Define paths, number of classes, and class names.

```python

data_cfg = {
    'path': config.YOLO_DATA_DIR,
    'train': "images/train",
    'val': "images/val",
    'nc': 1,         # number of classes
    'names': ['motor']  # class names
}

with open(f'{config.SRC}/data.yaml', 'w') as f:
    yaml.dump(data_cfg, f)

print('data.yaml written successfully')
```

## 3. (Optional) Custom Hyperparameters (`hyp.yaml`)
Adjust settings to handle class imbalance (focal loss, mosaic, mixup).

```python

hyp_params = {
    'lr0': 0.01,
    'box': 0.05,
    'obj': 1.0,
    'cls': 0.5,
    'fl_gamma': 1.5,
    'label_smoothing': 0.0,
    'mosaic': 1.0,
    'mixup': 0.5
}

with open(f'{config.SRC}/hyp.yaml', 'w') as f:
    yaml.dump(hyp_params, f)

print('hyp.yaml written successfully')
```

## 4. Train the YOLOv8 Model
Use the Ultralytics `YOLO` class to start training.

```python

# Load a pretrained YOLOv8 model
model = YOLO('yolov8n.pt')
print(f'{config.SRC}/data.yaml')

results = model.train(
    data=f'{config.SRC}/data.yaml',           # dataset config path
    epochs=7,                  # total epochs (int)
    batch=16,                   # batch size (int or float for auto)
    imgsz=640,                  # image size (int or list)
    device='cpu',  
    augment = True, 
    workers=8,                  # dataloader workers (int)
    optimizer='SGD',            # optimizer: 'auto', 'SGD', 'Adam', etc.
    lr0=0.01,                   # initial learning rate
    lrf=0.01,                   # final lr = lr0 * lrf
    momentum=0.937,             # momentum for SGD or beta1 for Adam
    weight_decay=0.0005,        # L2 regularization
    patience=10,                # early stopping patience (int)
    save=True,                  # save checkpoints
    save_period=-1,             # checkpoint frequency (epochs)
    cache=False,                # cache mode: False, 'ram', or 'disk'
    project=config.YOLO_RESULT,       # save results to runs/train/
    name='motor_detection',     # run name
    exist_ok=True,              # overwrite existing outputs
    pretrained=True,            # use pretrained weights
    seed=42,                    # random seed
    deterministic=True,         # deterministic training
    single_cls=False,           # single-class training
    rect=False,                 # rectangular training
    multi_scale=False,          # multi-scale training
    cos_lr=False,               # cosine LR scheduler
    close_mosaic=10,            # disable mosaic in final epochs
    resume=False,               # resume from last checkpoint
    amp=True,                   # automatic mixed precision
    warmup_epochs=3.0,          # warmup epochs
    warmup_momentum=0.8,        # warmup momentum
    warmup_bias_lr=0.1,         # warmup bias lr
    box=7.5,                    # box loss gain
    cls=0.5,                    # classification loss gain
    dfl=1.5,                    # distribution focal loss gain
    val=True,                   # validate during training
    plots=False                 # save plots
)

print(results)
```
