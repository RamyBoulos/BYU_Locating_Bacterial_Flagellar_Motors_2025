```python
import os
import sys
from ultralytics import RTDETR
import yaml
import torch


# Add project root to system path (for relative imports to work)
project_root = os.path.abspath(os.path.join(os.getcwd(), "..", ".."))
if project_root not in sys.path:
    sys.path.insert(0, project_root)

    
from src import config

```

```python
data_cfg = {
    'path': config.YOLO_DATA_DIR,
    'train': "images/train",
    'val': "images/val",
    'nc': 1,         # number of classes
    'names': ['motor']  # class names
}

with open(f'{config.SRC}/data.yaml', 'w') as f:
    yaml.dump(data_cfg, f)

print('data.yaml written successfully')
```

```python

model = RTDETR("rtdetr-l.pt")
results = model.train(
    data=f'{config.SRC}/data.yaml',
    project=config.RTDETR_RESULT,
    name='motor_detection_test',
    epochs=10,             
    batch=16,
    imgsz=640,
    device= torch.device("cuda" if torch.cuda.is_available() else "cpu"),
    optimizer="AdamW",       
    lr0=1e-4,                
    lrf=0.1,
    patience=20,
    cache=True,              
    #  augmentations
    augment=True,
    hsv_h=0.0, hsv_s=0.1, hsv_v=0.2,
    degrees=10, translate=0.1, scale=0.15, shear=0.0,
    fliplr=0.5, flipud=0.5, perspective=0.0,
    mosaic=0.0, mixup=0.0, cutmix=0.0,   # medikalde agresif augment kapalı
    # save/output settings
    save_period=1,
    verbose=True,
)
print(results)
```
